```python
#Label Encoder -> it converts the categorical values into numerical values.


from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
for i in df.columns:
  if df[i].dtypes == 'object':
    df[i] = le.fit_transform(df[i])


  # fit -> learn the pattern
# transform ->



# amit  -> 0
# anurag  -> 1
# 2 3 4 5 6 7
```

```python
# iloc -> index based fetching
# loc -> label based fetching
```

```python
x = df.iloc[:,:-1]
y = df['y']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 0)
```

```python
final_model = LogisticRegression()
```

```python
final_model.fit(x_train, y_train) # Basically learn from training data with given input and output

```

```python
y_pred = final_model.predict(x_test)
```

```python
y_pred
```

```python
y_pred.shape
```

```python
#in evaluation we checks the accuracy score (logistic -> accuracy)
from sklearn.metrics import *
```

```python
import sklearn.metrics as sp
```

```python
# Categorial Data  -> accuracy, precision, recall, f1-score, ROC-AUC
# Numerical Data -> rmse, mae, r^2

print("Accuracy score", accuracy_score(y_test, y_pred)* 100 )
print("Precision score", precision_score(y_test,y_pred) * 100 )
```

```python
print("Recall score is:",recall_score(y_test,y_pred)*100)
print("F1 score is:",f1_score(y_test,y_pred)*100)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

ConfusionMatrixDisplay(confusion_matrix=cm).plot(cmap="Blues")
plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

```python
print(classification_report(y_test, y_pred))
```

```python
# Model 1: With duration (existing model)
accuracy_with = accuracy_score(y_test, y_pred)
precision_with = precision_score(y_test, y_pred)
recall_with = recall_score(y_test, y_pred)
f1_with = f1_score(y_test, y_pred)
```

```python
# Model 2: Without duration
x_train_without_duration = x_train.drop(columns=["duration"])
x_test_without_duration = x_test.drop(columns=["duration"])

model_without_duration = LogisticRegression()
model_without_duration.fit(x_train_without_duration, y_train)

y_pred_without_duration = model_without_duration.predict(
    x_test_without_duration
)
```

```python
accuracy_without = accuracy_score(y_test, y_pred_without_duration)
precision_without = precision_score(y_test, y_pred_without_duration)
recall_without = recall_score(y_test, y_pred_without_duration)
f1_without = f1_score(y_test, y_pred_without_duration)
```

```python
comparison = pd.DataFrame({
    "With Duration": [
        accuracy_with,
        precision_with,
        recall_with,
        f1_with
    ],
    "Without Duration": [
        accuracy_without,
        precision_without,
        recall_without,
        f1_without
    ]
}, index=["Accuracy", "Precision", "Recall", "F1 Score"])

display(comparison.round(4))
```

```python
comparison.plot(kind="bar", figsize=(10, 5), rot=0)

plt.title("Model Performance: With vs Without Duration")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```
